## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### ML Task Type

This problem is a **Ranking / Scoring** task.

The goal is to assign every webpage a priority score based on its search performance, engagement, freshness, and other SEO signals.

Pages with higher scores should be reviewed and refreshed before lower-scoring pages.

Ranking is appropriate because the business needs an ordered list of webpages rather than a simple Yes/No prediction.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target / Proxy

The target for this ML task is a **Refresh Priority Score (Content Opportunity Score)** assigned to each webpage.

Since the dataset does not contain a direct "needs refresh" label, the model uses observed SEO and engagement signals such as impressions, clicks, sessions, average search position, engagement rate, and trend metrics as proxy indicators.

The output is a score that ranks webpages from highest to lowest refresh priority, helping the content team decide which pages should be reviewed first.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success Metric

The success metric for this task is **Precision@K**.

Since the goal is to rank webpages by refresh priority, the content team will review only the highest-ranked pages. Precision@K measures how many of the top recommended webpages are truly good candidates for content refresh.

A higher Precision@K means the model is placing the most valuable refresh opportunities near the top of the ranking, helping the team spend their time on the pages that matter most.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why Machine Learning?

A simple rule-based system is not enough for this problem because webpage performance depends on many factors rather than a single metric.

For example, a page with high impressions but a low click-through rate, declining traffic, and poor engagement may deserve a higher refresh priority than a page with low impressions but stable performance.

Machine Learning can analyze multiple SEO and engagement signals together, identify complex patterns, and assign a more accurate refresh priority score. This helps the content team focus on the webpages that are most likely to benefit from updates and improves decision-making compared to using fixed rules.
